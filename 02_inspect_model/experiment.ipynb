```python
import os

custom_cache_dir = "./hugging_face_cache"

# Create the folder if it doesn't exist yet
if not os.path.exists(custom_cache_dir):
    os.makedirs(custom_cache_dir)

# Safely set the environment variable
os.environ["HF_HOME"] = custom_cache_dir
```

```python
from transformers.pipelines import pipeline
```

```python
from dotenv import load_dotenv
load_dotenv()

hf_token = os.environ["HF_TOKEN"]
if hf_token:
    print("token access successfully")
```

```python
text_generator = pipeline(
    task = "text-generation",
    model="meta-llama/Meta-Llama-3.1-8B-Instruct",
    token=hf_token
)
```

```python
response = text_generator(text_inputs="Why is the color of sky?")
response
```

### Access the downloaded weights

```python
## Step 1: Only access the weights

from safetensors.torch import load_file
import os

# Path to one of the weight shards in your cache folder
shard_path = "./hugging_face_cache/hub/models--meta-llama--Meta-Llama-3.1-8B-Instruct/snapshots/0e9e39f249a16976918f6564b8830bc894c89659/model-00001-of-00004.safetensors"

# Change backend to "file" to stop Windows pagefile allocation
weights = load_file(shard_path, backend="file")

for name, tensor in list(weights.items())[:3]:
    print(f"Layer Name: {name} | Numbers inside: \n{tensor}\n")

```

```python
import os
from transformers import AutoModelForCausalLM

# Ensure your custom path environment variable is still set
os.environ["HF_HOME"] = "./hugging_face_cache"

# Pass the clean model identifier. Transformers handles the cache folder lookup automatically!
model = AutoModelForCausalLM.from_pretrained(
    "meta-llama/Meta-Llama-3.1-8B-Instruct", 
    device_map="cpu"
)

# View the weights safely
print(model.state_dict()["lm_head.weight"])
```
